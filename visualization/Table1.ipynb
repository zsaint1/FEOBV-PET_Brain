```python
#######  Table 1 stats
import pandas as pd
from tableone import TableOne

## Load data
df = pd.read_csv('data/masterfile_final_18patients_and_33controls.csv')

## Define variables/columns
columns = ['dem_age', 'dem_sex', 'smoking_status', 'injected_dose', 'IQ', 'CANTAB_composite_score', 'doi','DBI', 'cz_eq', 'Marder_Positive', 'Marder_Negative', 'Marder_Disorganized', 'Marder_Hostility', 'Marder_DepAnx', 'total_score_panss']
categorical = ['dem_sex', 'smoking_status']
groupby = 'Group'

## Create table 1
table1 = TableOne(df, columns=columns, categorical=categorical, groupby=groupby, pval=True, htest_name=True)

# Save table 1 as CSV
table1.to_csv('data/table1_18patients_33controls.csv')

## Display
table1
```
